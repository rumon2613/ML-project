21. Does this tiny dataset prove that snack purchases cause failure?

No. This dataset does not prove that snack purchases cause failure.

For example, Chayan bought 5 snacks and failed, but Emon bought 4 snacks and also failed. However, that does not establish that snacks caused their failure.

There are several reasons:

Only 6 students are included.
This is observational data, not a controlled experiment.
Other factors such as study hours, attendance, and sleep may affect the result.
Correlation does not necessarily mean causation.
A much larger and properly designed study would be needed to investigate whether snack purchases are associated with academic outcomes.

* Run the cell twice. Record which outputs remain identical. Then explain: if two groups claim different
accuracies because they used different random splits, can we fairly compare their models?

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)

    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```

Same model + different random split → different accuracy can occur.


Same split + same evaluation procedure → much fairer model comparison.
